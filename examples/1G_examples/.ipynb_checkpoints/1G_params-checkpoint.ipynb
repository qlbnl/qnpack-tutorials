## Few Network and Trap Parameters

- **retries**: The maximum number of attempts allowed to establish entanglement over a single network segment before declaring failure.  

- **coherence_time (ns)**: The duration for which a qubit can retain its quantum state before losing coherence due to environmental noise.  

- **init_photon_loss (QFC loss)**: The probability that a photon remains intact after undergoing quantum frequency conversion for transmission.

- **photon_loss (fibre loss)**: The photon survival probability per channel length [dB/km]. 

- **proto_sched (control protocols)**: The method of managing entanglement operations, either locally at each node (*hop-by-hop*) or through a *centralized controller*.  

- **Fidelity**: A measure of how accurately the final shared entangled state matches the ideal target state.  

- **Rate**: The number of successfully generated entangled pairs per unit time in a quantum network.

### Varying number of retries as well as distance for a single repeater network

```python
from qnpack.oneG.iontrap import IonTrapSimulation
fixed_params = {
    "num_repeaters": 1,
    "photon_loss": 0.2,
    "coherence_time": 60000000, #ns
}

varying_params = {
    "retries": [30, 60, 90],
    "distance": [10, 20, 30, 40], #km
}

sim = IonTrapSimulation(fixed_params=fixed_params, varying_params=varying_params,
                        iterations=10,
                        parameter_file="parameters/noisy.yml",
                        output_dir="results")
data = sim.start()
sim.plot(final_data=data, x_axis2="distance", y_axis2="fidelity", xlabel2="Distance (km)", ylabel2="Fidelity",
         label_param2="retries", title2="Fidelity vs Distance",
         x_axis1="distance", y_axis1="rate", xlabel1="Distance (km)", ylabel1="Rate (Hz)",
         label_param1="retries", title1="Rate vs Distance")
```
