# First 1G Repeater Network

This first example will demonstrate how to use QNPack to simulate a quantum network repeater chain using a single quantum repeater (QR).

<img src="./images/1g_network.png" width="75%"><br>


Key components modeled in the 1G Trapped Ion repeater chain network:

- Q-Nodes
- BSM-Nodes
- Repater-Nodes
- Control-Nodes
- Classical and Quantum channels

### Varying network length and number of repeaters in a noise-less network

```python
from qnpack.oneG.iontrap import IonTrapSimulation
fixed_params = {}

varying_params = {
    "num_repeaters": [1, 3, 5],
    "distance": [5, 10, 15],
}

sim = IonTrapSimulation(fixed_params=fixed_params, varying_params=varying_params,
                        iterations=10,
                        parameter_file="parameters/noiseless.yml",
                        output_dir="results")
data = sim.start()
sim.plot(final_data=data, x_axis1="distance", y_axis1="fidelity", xlabel1="Distance (km)", ylabel1="Fidelity",
         label_param1="num_repeaters", title1="Fidelity vs Distance",
         x_axis2="distance", y_axis2="rate", xlabel2="Distance (km)", ylabel2="Rate (Hz)",
         label_param2="num_repeaters", title2="Rate vs Distance")
```

### Varying network length and number of repeaters in a noisy network

```python
from qnpack.oneG.iontrap import IonTrapSimulation
fixed_params = {
    "photon_loss": 0.2,
    "coherence_time": 60000000, #ns
}

varying_params = {
    "num_repeaters": [1, 3, 5],
    "distance": [5, 10, 15],
}

sim = IonTrapSimulation(fixed_params=fixed_params, varying_params=varying_params,
                        iterations=10,
                        parameter_file="parameters/noisy.yml",
                        output_dir="results")
data = sim.start()
sim.plot(final_data=data, x_axis1="distance", y_axis1="fidelity", xlabel1="Distance (km)", ylabel1="Fidelity",
         label_param1="num_repeaters", title1="Fidelity vs Distance",
         x_axis2="distance", y_axis2="rate", xlabel2="Distance (km)", ylabel2="Rate (Hz)",
         label_param2="num_repeaters", title2="Rate vs Distance")
```
