# First All-photonic Repeater Network

This first example will demonstrate how to use QNPack to simulate a quantum network repeater chain using a single quantum repeater (QR).

<img src="./images/ape_network.png" width="75%"><br>

```python
from qnpack.APE.allphotonic import APESimulation

sim = APESimulation(num_repeaters=[1,3,5],
                    distances=[10],
                    iterations=1,
                    min_successful=10,
                    parameter_file="parameters/scenario1.yml",
                    output_dir="results")
sim.start()
```
