# QNPack: Modeling and Simulation of Quantum Repeater Networks

This tutorial environment comes with a number of examples for 1G and All-Photonic quantum repeater experiments.

These examples run in a short amount of time to showcase the usage of the QNPack modules.

## Getting started

- Setup Environment

  - __JupyterHub__ (JH)
    
    The QNPack software is pre-installed with all necessary dependencies inside the JH container image. If working inside JH, the examples notebooks below will work without any additional setup.
  - __Source Install__
  
    Installing QNPack from source requires a number of dependencies, including NetSquid.

    To install NetSquid, register at http://netsquid.org and install the following packages:
    ```bash
    pip3 install --extra-index-url https://pypi.netsquid.org netsquid netsquid-trappedions netsquid-netconf
    ```

    Then, inside the QNPack source directory:
    ```bash
    pip3 install -r requirements.txt
    ```
