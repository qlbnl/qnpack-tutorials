```python
import matplotlib.pyplot as plt
import networkx as nx

def draw_network(n_repeaters, distance_km, title):
    G = nx.Graph()
    
    # Add Q-nodes
    G.add_node("QNode_A", type="QNode")
    G.add_node("QNode_B", type="QNode")
    
    # Add Repeaters and BSM nodes
    for i in range(n_repeaters):
        G.add_node(f"Repeater_{i+1}", type="Repeater")
    for i in range(n_repeaters + 1):
        G.add_node(f"BSM_{i+1}", type="BSM")
    
    # Create edges
    G.add_edge("QNode_A", "BSM_1")
    G.add_edge("BSM_1", "Repeater_1")
    for i in range(n_repeaters - 1):
        G.add_edge(f"Repeater_{i+1}", f"BSM_{i+2}")
        G.add_edge(f"BSM_{i+2}", f"Repeater_{i+2}")
    G.add_edge(f"Repeater_{n_repeaters}", f"BSM_{n_repeaters+1}")
    G.add_edge(f"BSM_{n_repeaters+1}", "QNode_B")
    
    # Define straight horizontal positions
    num_nodes = 2 * n_repeaters + 3
    spacing = 1.0 / (num_nodes - 1)
    
    pos = {}
    pos["QNode_A"] = (0, 0)
    x_pos = spacing
    for i in range(n_repeaters + 1):
        pos[f"BSM_{i+1}"] = (x_pos, 0)
        x_pos += spacing
        if i < n_repeaters:
            pos[f"Repeater_{i+1}"] = (x_pos, 0)
            x_pos += spacing
    pos["QNode_B"] = (1, 0)
    
    # Assign colors based on actual node names
    node_colors = []
    for node in G.nodes():
        if node in ["QNode_A", "QNode_B"]:
            node_colors.append("lightblue")   # First and last QNodes
        elif node.startswith("Repeater"):
            node_colors.append("orange")      # Repeaters
        elif node.startswith("BSM"):
            node_colors.append("lightgreen")  # BSM nodes
        else:
            node_colors.append("grey")
    
    # Draw the network
    plt.figure(figsize=(12, 2))
    nx.draw(G, pos, with_labels=True, node_color=node_colors, node_size=1000, font_size=8)
    
    plt.title(title, fontsize=14)
    plt.axis('off')
    plt.show()

# Example
draw_network(4, 400, "Quantum Repeater Network")
```
